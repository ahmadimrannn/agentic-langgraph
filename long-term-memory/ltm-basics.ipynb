```python
from langgraph.store.memory import InMemoryStore
from huggingface_hub import InferenceClient
import os
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
embedding_model = InferenceClient(model="sentence-transformers/all-MiniLM-L6-v2", token=os.getenv("HF_TOKEN"))

def embed(texts: list[str]) -> list[list[float]]:
    return embedding_model.feature_extraction(texts).tolist()

store = InMemoryStore(index={"embed": embed, "dims": 384})
```

```python
namespace = ("users", "u1")
```

```python
store.put(namespace, "1", {"data": "Name: the name of the user is Ahmad."})
store.put(namespace, "2", {"data": "The user likes biryani, pizza, and burgers from Burger's War."})
store.put(namespace, "3", {"data": "The user is a Muslim."})
store.put(namespace, "4", {"data": "The user is interested in artificial intelligence and machine learning."})
store.put(namespace, "5", {"data": "The user is learning to build AI agents using LangGraph."})
store.put(namespace, "6", {"data": "The user is interested in becoming an Agentic AI Engineer."})
store.put(namespace, "7", {"data": "The user uses Python for programming."})
store.put(namespace, "8", {"data": "The user is interested in career development and technology."})
store.put(namespace, "9", {"data": "The user uses VS Code and Jupyter notebooks for development."})
store.put(namespace, "10", {"data": "The user prefers practical, hands-on explanations when learning technical concepts."})
```

```python
namespace2 = ("users", "u2")
```

```python
store.put(namespace2, "1", {"data": "Name: The name of the user is Sohaib."})
store.put(namespace2, "2", {"data": "the user is a muslim."})
store.put(namespace2, "3", {"data": "The user really likes to eat spicy foods."})
store.put(namespace2, "4", {"data": "The user is learning full stack development"})
```

```python
search_results = store.search(namespace, limit=1000)

exact_items_count = len([
    item for item in search_results 
    if item.namespace == namespace
])

for i in range(1, exact_items_count + 1):
    item = store.get(namespace, f"{i}")
    print(item.value['data'])
```

```python
items = store.search(namespace)

for item in items:
    print(item.value['data'])
```
